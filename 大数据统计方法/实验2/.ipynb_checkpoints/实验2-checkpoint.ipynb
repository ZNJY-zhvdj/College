行类这么些$\sum x_i$okl

```python
def a_list(lst):
    s = [x for x in lst if x % 2 == 0]
    print(s)

   
    b_lst = lst[::-1]
    print(b_lst)

    cs = [x for x in lst if x % 2 != 0]
    c_dict = {i: cs[i] for i in range(len(cs))}
    print(c_dict)


my_list = list(range(100))

a_list(my_list)
```
