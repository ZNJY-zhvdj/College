```python
student['身高 (m)'] = student['身高 (cm)'] / 100
student
```

```python
student['5级成绩'] = pd.cut(
    student['考试成绩'],
    bins=[-np.inf, 60, 70, 80, 90, np.inf],
    labels=['不及格', '及格', '中等', '良好', '优秀'],
    right=False   
)

#pd.cut(x, bins, right=True, labels=None,)
#x为要分箱的数据，bins是分箱的边界
#right=true默认(a, b]，right=false为[a,b]，labels是给区间起名字
#np.inf表示无穷

student
```

```python
student[student['姓名'].str.startswith('李')]

#df[df['列名'].str.方法(参数)]
```

```python
student['身高_带单位'] = student['身高 (cm)'].astype(str)+ '厘米'
student['体重_带单位'] = student['体重 (kg)'].astype(str)+ '千克'
student

#.astype(str)是转换成字符串
```

```python
student.groupby('性别').agg(
    平均成绩=('考试成绩', 'mean'),
    平均身高=('身高 (cm)', 'mean')
)
```

```python
student.groupby('籍贯').agg(
    学生数量=('姓名','count'),
    平均成绩=('考试成绩','mean'),
    平均体重=('体重 (kg)','mean')
)
```

```python
class1=pd.read_excel('教学信息.xlsx',sheet_name='班级')
class1
```

```python
class1_2 = class1.merge(teacher,
    left_on='班主任工号',  
    right_on='工号',
    how='left' 
)
class1_2['班主任工号'] = class1_2['姓名']
class1_2


#pd.merge() 的核心口诀：
#左表.merge(右表, left_on='左表键', right_on='右表键', how='连接方式')
```

```python
xk=pd.read_excel('教学信息.xlsx',sheet_name='选课')
xk
```

```python
xk["学生姓名"] = xk["学号"].map(student.set_index("学号")["姓名"])
xk["班级名称"] = xk["班级编号"].map(class1_2.set_index("班级编号")["班级名称"])
xk["教师姓名"] = xk["教师编号"].map(teacher.set_index("工号")["姓名"])
xk



#xk[新列名] = xk[键列名].map(映射表)
#映射表student.set_index("键列名")["值列名"]
```
