```python
b1 = np.arange(1,13).reshape(3,4)
```

```python
print(b1)
```

```python
print(b1.max())
print(b1.min())
print(b1.sum())
print(b1.mean())

# 中位数需要用 np.median
print(np.median(b1))

 # 方差需要用 np.var
print(np.var(b1))

print(b1.max(axis=1))
print(b1.min(axis=1))
#axis=0 和 axis=1 是新手最容易晕的地方。你可以这样记：axis=0 代表“往下走”（跨行，压扁行，得到列结果）；axis=1 代表“往右走”（跨列，压扁列，得到行结果）。
print(b1.max(axis=0))
print(b1.min(axis=0))
```

```python
c = np.arange(1,26).reshape(5,5)
print(c)
```

```python
# 切片语法 [行起点:行终点, 列起点:列终点]，取前不取后
print(c[0:3, 0:3])
print(c[0:3, 2:5])
print(c[2:5, 0:3])
print(c[2:5, 2:5])

print(c[1:4, 1:4])
```

```python
d = np.array([[c[0, 0], c[0, -1]],
                    [c[-1, 0], c[-1, -1]]])
print(d)
```

```python
# 4. 计算这个5x5矩阵的每个行上的平方和
# 先让矩阵每个元素平方 (c**2)，然后按行求和 (axis=1)
print("每行平方和：", np.sum(c ** 2, axis=1))
#np.sum(..., axis=1) 的意思是：沿着行（水平方向）把数字加起来，最终得到每一行的总和。


# 5. 计算这个5x5矩阵的每个列上的平方和
# 先让矩阵每个元素平方 (c**2)，然后按列求和 (axis=0)
print("每列平方和：", np.sum(c ** 2, axis=0))
```

```python
v1 = np.array([1, 1])
v2 = np.array([1, 0])

# 计算点积（内积）
dot = np.dot(v1, v2)
# 计算模长
norm_v1 = np.linalg.norm(v1)
norm_v2 = np.linalg.norm(v2)

# 计算余弦值 (加上 np.clip 防止浮点误差报错)
cos_theta = np.clip(dot / (norm_v1 * norm_v2), -1.0, 1.0)

# 求弧度与角度
rad = np.arccos(cos_theta)
deg = np.degrees(rad)

print("夹角（弧度）：", rad)
print("夹角（角度）：", deg)
```
